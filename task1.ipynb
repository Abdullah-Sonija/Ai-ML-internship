```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
print(f"Shape: {df.shape}")
```

```python
print(f"Column Names: {df.columns.tolist()}")
```

```python
sns.set_theme(style="whitegrid")
fig, axs = plt.subplots(1, 2, figsize=(15, 7))

sns.scatterplot(data= df, x= 'sepal_length', y= 'sepal_width', hue= 'species', palette= 'deep', s= 100, ax= axs[0])
axs[0].set_title('Scatter Plot: Sepal Length vs. Sepal Width', fontsize= 14)

sns.boxplot(data= df, ax= axs[1], orient= "h")
axs[1].set_title('Box Plots: Identifying Spread and Outliers', fontsize=14)
axs[1].set_xlabel('Centimeters (cm)')

fig.tight_layout()
```

```python
df.hist(figsize=(11,8), bins= 20, edgecolor= "black", color= "pink")
plt.suptitle("Histograms: Value Distribution of Features", fontsize= 15)
plt.tight_layout()
```

```python
sns.pairplot(df, hue= "species", palette="bright")
plt.tight_layout()
```
